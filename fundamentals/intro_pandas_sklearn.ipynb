## 1. Import the libraries

The conventional alias for Pandas is `pd`.

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

## 2. Load the CSV file

`pd.read_csv()` reads the CSV file and stores it in a Pandas DataFrame.

The CSV file should be stored in the same folder as this notebook.

```python
df = pd.read_csv("iris_intro.csv")
df
```

## 3. View the first rows

`head()` displays the first five rows by default.

```python
df.head()
```

## 4. Inspect the DataFrame

`info()` displays the column names, data types, and number of non-empty values.

```python
df.info()
```

## 5. View summary statistics

`describe()` calculates useful statistics for numerical columns.

```python
df.describe()
```

## 6. Select one column

A single column can be selected using its name.

```python
df["sepal_length_cm"]
```

## 7. Select several columns

A list of column names can be used to select more than one column.

```python
df[["sepal_length_cm", "petal_length_cm", "species"]]
```

## 8. Filter rows

The following code selects flowers with a petal length greater than 5 cm.

```python
large_petals = df[df["petal_length_cm"] > 5]
large_petals.head()
```

## 9. Calculate a mean

The following code calculates the mean sepal length.

```python
mean_sepal_length = df["sepal_length_cm"].mean()
print(f"Mean sepal length: {mean_sepal_length:.2f} cm")
```

## 10. Count categories

`value_counts()` counts the number of rows for each flower species.

```python
df["species"].value_counts()
```

## 11. Create a new column

A new column can be created from existing columns.

```python
df["sepal_area"] = df["sepal_length_cm"] * df["sepal_width_cm"]
df.head()
```

# Introduction to machine learning

The task is to predict a flower's species from four measurements.

The four measurement columns are called the **features**.

The `species_id` column is called the **target**.

## 12. Select the features and target

```python
feature_columns = [
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm",
]

X = df[feature_columns]
y = df["species_id"]

print("Feature data shape:", X.shape)
print("Target data shape:", y.shape)
```

## 13. Split the data

The training data is used to teach the model.

The test data is kept separate and used to evaluate the model.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y,
)

print("Training rows:", len(X_train))
print("Test rows:", len(X_test))
```

## 14. Create and train a decision tree

A decision tree is a simple classification model.

`random_state` makes the result reproducible.

```python
model = DecisionTreeClassifier(
    max_depth=3,
    random_state=42,
)

model.fit(X_train, y_train)
```

## 15. Make predictions

```python
predictions = model.predict(X_test)
predictions[:10]
```

## 16. Measure accuracy

Accuracy is the proportion of correct predictions.

```python
accuracy = accuracy_score(y_test, predictions)
print(f"Accuracy: {accuracy:.2%}")
```

## 17. Compare actual and predicted values

```python
results = pd.DataFrame({
    "actual_species_id": y_test.to_numpy(),
    "predicted_species_id": predictions,
})

results.head(10)
```

## 18. Predict one new flower

The new flower must contain the same four measurements, in the same order as the training data.

```python
new_flower = pd.DataFrame(
    [[5.1, 3.5, 1.4, 0.2]],
    columns=feature_columns,
)

predicted_id = model.predict(new_flower)[0]

species_names = {
    0: "setosa",
    1: "versicolor",
    2: "virginica",
}

print("Predicted species:", species_names[predicted_id])
```

Try a different classifier and compare its accuracy with the decision tree.
